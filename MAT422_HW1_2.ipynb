1.2.1 Linear Spaces
A vector space is a set of vectors that is closed under vector addition and scalar multiplication.

This means that if two vectors belong to the same vector space, their sum must also belong to the space as well as scalar multiplication.

```python
import numpy as np
## Vector addition and scalar multiplication
## Going to show that adding vectors or multiplying scalars remain in R^3


u = np.array([1,2,3])
v = np.array([4,-1,2])
c = 3

vector_sum = u + v
scalar_multiple = c * u

print("u = ",u)
print("v = ",v)
print("u + v =", vector_sum)
print("3u =", scalar_multiple)
```

The results are vectors in the same space R^3.


```python
## Zero vector
## A Vector space contains the zero vector, adding this to a vector
## does not change the vector

zero = np.array([0,0,0])

print("u + 0 = ", u + zero)
print(" u =", u)


```

Adding the zero vector to vector u gives u again.

```python
## Additive Inverse
## Every vector has an additive inverse. A vector added
## to its inverse gives the zero vector.

neg_u = -u
print(" -u =", neg_u)
print("u + (-u) =", u + neg_u)
```

The vector "-u" is the additive inverse of u because their sum is the zero vector.

```python
## Linear Combination and Span
## A linear combination is formed by multiplying vectors
## by scalars and adding the results.

w = 2 * u - v
print("w = 2u - v ", w)
```

The vector w is a linear combination of u and v. Therefore, w belongs to the span of u and v.

1.2.2 Orthogonality

Two vectors are orthogonal if their dot product is equal to zero.

for vectors u and v:

u dot v = 0

Orthogonal vectors meet at a 90 degree angle.

```python
## Dot product

u = np.array([1,2])
v = np.array([2,-1])

print("u =", u)
print("v =", v)

dot_product = np.dot(u,v)
print("u dot v = ", dot_product)
```

```python
## Geometric interpretation
## Norm and angle between vectors

## Norm represents vectors length
## Angle between vectors can be found using
## cos(theta) = (u * v) / (||u|| ||v||)


#norm
norm_u = np.linalg.norm(u)
norm_v = np.linalg.norm(v)

print("||u|| =", norm_u)
print("||v|| =", norm_v)
```

```python
## Angle between vectors
cos_theta = dot_product / (norm_u * norm_v)

theta = np.arccos(cos_theta)
theta_degrees = np.degrees(theta)

print("cos(theta) =", cos_theta)
print("theta = ", theta_degrees, "degrees")
```

The angle is 90 degrees so they are orthogonal

```python
## Projection
## The projection of u onto v measures how much of
## u points in the direction of v

projection = (np.dot(u,v) / np.dot(v,v)) * v
print("Projection of u onto v = ", projection)
```

The projection is the zero vector because u and v are orthogonal. Neither vector has a component in the direction of the other.

The projection produced an output of -0, it should be 0 but it produced a number so small because of floating point roundoff.

```python
## Gram-Scmidt Process
## This process takes linearly independent vectors
## and converts them into an orthonormal set

## It works by removing the projection of one vector onto
## another then normalizig the result

v1 = np.array([1.0,1.0])
v2 = np.array([1.0,0.0])
```

Step 1: First vector is kept as the first orthogonal vector

```python
u1 = v1
print("u1 =", u1)
```

Step 2: Remove the projection to make the second vector orthogonal to the first, subtract the projection of v2 onto u1.

```python
projection = (np.dot(v2,u1)/ np.dot(u1,u1))* u1

u2 = v2 - projection

print("Projection of v2 onto u1 =",projection)
print("u2 =", u2)
```

If u1 and u2 are orthogonal then dot product should be zero

```python
print("u1 dot u2 =", np.dot(u1,u2))
```

Now to normalize the vectors to get them to be length = 1

```python
q1 = u1 / np.linalg.norm(u1)
q2 = u2 / np.linalg.norm(u2)

print("q1 =", q1)
print("q2 =", q2)
```

```python
print("||q1|| = ", np.linalg.norm(q1))
print("||q2|| = ", np.linalg.norm(q2))
```

Norm is basically 1

```python
# Verifying result
Q = np.column_stack((q1,q2))

print("Q =")
print(Q)

print("\nQ.T @ Q =")
print(Q.T @ Q)
```

The -2 diagonal entries are basically zero so we have the identity matrix.

1.2.4 Eigenvalues and Eigenvectors

An eigenvector is a direction that a matrix changes only by scaling.

For a matrix A, an eigenvector v and eigenvalue lamba satisfy:

A * v = lamba * v

```python
A = np.array ([
    [2,1],
    [1,2]
])

print("A =")
print(A)
```

```python
eigenvalues, eigenvectors = np.linalg.eig(A)

print("Eigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)
```

To verify an eigenvector and eigenvalue, I will check whether multiplying the eigenvector by A gives the same result as multiplying the eigenvector by its eigenvalue.

```python
lam = eigenvalues[0]
v = eigenvectors[:,0]

Av = A @ v
lambda_v = lam * v

print("lambda = ", lam)
print("v = ", v)

print("\nA @ v =", Av)
print("lambda * v =", lambda_v)
```

```python
#Residual Norm
residual = Av - lambda_v
residual_norm = np.linalg.norm(residual)

print("Residual =", residual)
print("Residual norm =", residual_norm)
```

The residual norm is approximately 0, which confirms that A @ v is equal to lambda * v. Therefore, v is an eigenvector of A with eigenvalue lambda.

```python
v1 = eigenvectors[:,0]
v2 = eigenvectors[:,1]
print("Dot product of eigenvectors = ", np.dot(v1,v2))
```

Because A is a symmetric matrix, its eigenvectors are orthogonal. Their dot product is approx 0, confirming this property.

The eigenvectors represent directions that remain unchanged in direction when A is applied, while the eigenvalues represent how much those directions are scaled.
