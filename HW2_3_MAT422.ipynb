# 2.3.1 Joint Probability Distribution

A Joint probability Distribution describes the probability that two random variables take specific values at the same time.

Let X have possible values 0 and 1, and let Y have possible values 0, 1, and 2.

A valid joint probability distribution must have nonnegative probabilities and all probabilities must add up to 1.

We can also use joint distribution to calculate marginal and conditional probabilities.


```python
import numpy as np

joint = np.array([
    [0.15, 0.20, 0.10],
    [0.05, 0.20, 0.30]
])
x_values = np.array([0, 1])
y_values = np.array([0, 1, 2])

# Marginal Distributions
px = joint.sum(axis=1)
py = joint.sum(axis=0)

print("Joint distribution:")
print(joint)

print("\nJoint total =", joint.sum())
print("P(X) =", px)
print("P(Y) =", py)

# Conditional probability P(Y = 2 / X = 1)
conditional = joint[1, 2] / px[1]

print("\nP(Y=2 | X=1) =", conditional)

```

Summary

The joint probabilities for X and Y all add up to 1 so this is a valid joint probability distribution.

The conditional probability is approximately 54.55%. This means that when X=1, the probability that Y=2 is about 54.55%

# Correlation and Dependence

Correlation measures the strength and direction of a linear relationship between two variables.

In this example x is generated randomly. A dependent variable is create using x while another is generated seperately from x.

If two variables are independent we will see their correlation being closer to 0.

```python
import matplotlib.pyplot as plt
import numpy as np
rng = np.random.default_rng(422)

x = rng. normal(0,1,250)

# y_dependent is based on x plus some random noise
y_dependent = 1.5 * x + rng.normal(0,0.7, 250)

# y_independent is generated seperately from x
y_independent = rng.normal(0,1,250)

corr_dependent = np.corrcoef(x,y_dependent)[0,1]
corr_independent = np.corrcoef(x,y_independent)[0,1]

plt.scatter(x,y_dependent, label="Dependent")
plt.scatter(x,y_independent, label="Independent")
plt.legend()
plt.title("Correlation as a Dependence Summary")
plt.show()
```

Summary

The dependent variable has a positive correlation with x because it was constructed using x. Its points also show an upward trend in the scatter plot.

The independent variable has a correlation much closer to 0 because it was generated seperate from x. Its scatter plot does not show the same clear linear patter.

This demonstrates how correlation can summarize linear dependence between two variables.
